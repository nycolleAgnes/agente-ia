## Objetivo

Neste notebook, desenvolvemos um **Agente de Inteligência Artificial (IA)** capaz de interpretar automaticamente os resultados dos modelos de **Machine Learning** testados anteriormente — Regressão Logística e Random Forest.

O propósito é integrar o pipeline tradicional de aprendizado de máquina com uma camada explicativa em linguagem natural, aproximando a análise técnica de uma comunicação mais acessível e estratégica.

---

## Etapas principais

---

### 1. Configuração do ambiente

- O notebook é compatível tanto com Google Colab quanto com execução local (VS Code).
- Detecta automaticamente o ambiente de execução e define os caminhos base para leitura e salvamento dos resultados.
- Verifica e cria automaticamente a pasta **resultados/**, onde serão armazenados os arquivos gerados.

---

### 2. Simulação dos resultados de modelagem

- Criação de um *DataFrame* com as métricas obtidas pelos modelos ajustados anteriormente **(Regressão Logística e Random Forest)**.


- **As principais métricas consideradas são:**

  - Acurácia: taxa de acertos gerais;
  - F1-score: equilíbrio entre precisão e recall, importante para bases desbalanceadas.

- O arquivo é salvo como **resultados_modelos.csv** na pasta **resultados/**.

---

### 3. Agente de IA explicativo

- Implementação da função **agente_explicativo(resultados)** que:
- Identifica automaticamente o modelo com melhor F1-score;

- Gera uma explicação textual em linguagem natural, destacando:
- Qual modelo teve melhor desempenho;
- Diferenças entre os modelos;
- Interpretação dos resultados e sugestões de uso.

- O texto explicativo é impresso na tela e salvo em **resultados/explicacao_modelos.txt.**

---

### 4. Interpretação automática gerada pelo agente

- **O agente conclui, com base nas métricas simuladas, que:**
- A Regressão Logística apresenta melhor equilíbrio entre precisão e recall;
- O Random Forest, embora mais robusto, tem menor interpretabilidade.

- Essa explicação é redigida de forma compreensível, simulando o comportamento de um assistente inteligente capaz de justificar decisões de modelos.

---

### 5. Extensão futura — Integração com IA generativa real

- O notebook inclui um exemplo comentado mostrando como o agente pode futuramente se conectar a modelos de linguagem (como **OpenAI GPT, LLaMA ou Claude**) para gerar explicações ainda mais ricas e contextuais.
- Essa integração permitiria que o agente interpretasse resultados reais e sugerisse melhorias automáticas no pipeline de modelagem.

---

## Resultado esperado

Ao final da execução deste notebook, o usuário terá:

- Um arquivo CSV (**resultados_modelos.csv**) com as métricas dos modelos;
- Um arquivo de texto (**explicacao_modelos.txt**) com uma explicação automática gerada pelo agente;
- Um pipeline completo que une **Machine Learning tradicional e IA** explicativa, servindo como base para futuras integrações com agentes generativos reais.


```python
# 03_agente_ia.ipynb — Agente de IA para Interpretação de Modelos 

# 1️ Introdução
"""
Neste notebook, criamos um agente de IA que interpreta automaticamente os resultados
dos modelos de Machine Learning testados anteriormente (Regressão Logística e Random Forest).

Objetivo:
- Gerar análises automáticas em linguagem natural;
- Identificar o modelo mais adequado;
- Explicar de forma interpretável o desempenho e possíveis próximos passos.

Essa etapa conecta o pipeline de ML tradicional com um agente de IA explicativo,
demonstrando a integração entre Machine Learning e IA generativa.
"""

# 2️ Imports necessários
import pandas as pd
import os

# 3 Detecta ambiente: Colab ou local 
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')
    base_path = "/content/drive/MyDrive/agente-ia"
    print(" Executando no Google Colab.")
else:
    base_path = os.path.abspath("..")  # ajusta se o notebook estiver dentro de "notebooks/"
    print(" Executando localmente (VS Code).")

# Caminhos base e resultados
results_path = os.path.join(base_path, "resultados")
os.makedirs(results_path, exist_ok=True)
print(f" Pasta de resultados verificada/criada em: {results_path}")

# 4️ Resultados simulados (vindos da modelagem)
resultados = pd.DataFrame({
    "Modelo": ["Regressão Logística (Tuned)", "Random Forest (Tuned)"],
    "Acurácia": [0.5900, 0.6227],
    "F1-score": [0.4137, 0.3953]
})

display(resultados)

# Salva tabela de resultados
resultados_path = os.path.join(results_path, "resultados_modelos.csv")
resultados.to_csv(resultados_path, index=False)
print(f" Resultados salvos em: {resultados_path}")

# 5️ Agente de IA Explicativo
def agente_explicativo(resultados: pd.DataFrame) -> str:
    """
    Gera uma explicação em linguagem natural a partir das métricas dos modelos.
    """
    melhor_modelo = resultados.loc[resultados["F1-score"].idxmax(), "Modelo"]
    f1 = resultados.loc[resultados["F1-score"].idxmax(), "F1-score"]
    acc = resultados.loc[resultados["F1-score"].idxmax(), "Acurácia"]

    explicacao = f"""
    O modelo com melhor desempenho geral foi **{melhor_modelo}**,
    com F1-score de {f1:.3f} e acurácia de {acc:.3f}.

    Embora o Random Forest apresente uma acurácia ligeiramente superior,
    a Regressão Logística manteve melhor equilíbrio entre precisão e recall,
    o que a torna mais consistente para bases desbalanceadas.

    Em termos de interpretação, a Regressão Logística é linear e mais simples
    de explicar para stakeholders, enquanto o Random Forest é mais robusto,
    podendo capturar relações não lineares. Dependendo do objetivo de negócio,
    ambos podem ser usados: a regressão para previsões rápidas e interpretáveis,
    o Random Forest para cenários que exigem mais precisão.
    """
    return explicacao.strip()

# 6️ Executa o agente e salva explicação
explicacao_texto = agente_explicativo(resultados)
print("\n Explicação Gerada pelo Agente de IA:\n")
print(explicacao_texto)

# Salva a explicação em arquivo .txt
explicacao_path = os.path.join(results_path, "explicacao_modelos.txt")
with open(explicacao_path, "w", encoding="utf-8") as f:
    f.write(explicacao_texto)

print(f"\n Explicação salva em: {explicacao_path}")

# 7️ Extensão (comentada): Integração com IA Generativa real
"""
Futuramente, é possível integrar o agente com um modelo de linguagem (ex: OpenAI GPT, LLaMA, Claude etc.)
para gerar explicações mais detalhadas automaticamente. Exemplo:

from openai import OpenAI
client = OpenAI(api_key="SUA_CHAVE_API")

prompt = f"Explique os seguintes resultados de modelagem:\n{resultados.to_string()}"
resposta = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
print(resposta.choices[0].message.content)
"""

print("\n Pipeline finalizado com sucesso! Todos os resultados foram salvos com sucesso!")

```
